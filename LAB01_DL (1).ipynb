```python
import numpy as np

# Step 1: Training data (AND Gate)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target output
y = np.array([0, 0, 0, 1])

# Step 2: Initialize weights and bias
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 10

# Step 3: Activation function
def activation(x):
    if x >= 0:
        return 1
    else:
        return 0

# Step 4: Training the perceptron
for epoch in range(epochs):
    print(f"Epoch {epoch + 1}")
    for i in range(len(X)):
        # Calculate weighted sum
        net_input = np.dot(X[i], weights) + bias

        # Predict output
        prediction = activation(net_input)

        # Calculate error
        error = y[i] - prediction

        # Update weights and bias
        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error

        print(f"Input: {X[i]}, Target: {y[i]}, Prediction: {prediction}")

    print("Weights:", weights)
    print("Bias:", bias)
    print("-" * 30)

# Step 5: Testing
print("\nFinal Predictions:")
for i in range(len(X)):
    net_input = np.dot(X[i], weights) + bias
    prediction = activation(net_input)
    print(f"Input: {X[i]} --> Output: {prediction}")
```
