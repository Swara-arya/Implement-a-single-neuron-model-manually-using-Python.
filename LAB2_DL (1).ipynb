```python
import numpy as np

# Step 1: Input values
inputs = np.array([1, 2, 3])

# Step 2: Weights
weights = np.array([0.2, 0.4, 0.6])

# Step 3: Bias
bias = 0.5

# Step 4: Calculate weighted sum
weighted_sum = np.dot(inputs, weights) + bias

# Step 5: Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 6: Calculate output
output = sigmoid(weighted_sum)

# Step 7: Display results
print("Inputs:", inputs)
print("Weights:", weights)
print("Bias:", bias)
print("Weighted Sum:", weighted_sum)
print("Output:", output)
```
