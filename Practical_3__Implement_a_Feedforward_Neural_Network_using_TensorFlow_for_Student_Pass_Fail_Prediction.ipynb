Step 1: Import Required Libraries

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
```

Step 2: Prepare the Dataset

Features:

Study Hours
Attendance (%)
Assignment Score

Output:

1 → Pass

0 → Fail


```python
# Input Features
X = np.array([
    [2, 60, 45],
    [3, 65, 50],
    [5, 80, 70],
    [6, 85, 75],
    [7, 90, 85],
    [8, 95, 90],
    [1, 50, 35],
    [4, 75, 65]
], dtype=float)

# Target Output
y = np.array([
    0,
    0,
    1,
    1,
    1,
    1,
    0,
    1
], dtype=float)

print("Training Data Loaded Successfully!")
```

Step 3: Build the Feedforward Neural Network

The network contains:

Input Layer (3 neurons)

Hidden Layer (4 neurons)

Output Layer (1 neuron)


```python
model = Sequential()

# Hidden Layer
model.add(Dense(4, input_dim=3, activation='relu'))

# Output Layer
model.add(Dense(1, activation='sigmoid'))
```

Step 4: Compile the Model

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

Step 5: Train the Model

```python
model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)
```

Step 6: Test the Model

Suppose a student has:

Study Hours = 5

Attendance = 82%

Assignment Score = 72

```python
new_student = np.array([[5, 82, 72]])

prediction = model.predict(new_student)

print("Prediction Probability:", prediction[0][0])

if prediction >= 0.5:
    print("Result: PASS")
else:
    print("Result: FAIL")
```

Step 7: Test Multiple Students

```python
students = np.array([
    [2, 55, 40],
    [6, 88, 80],
    [3, 70, 55],
    [8, 96, 91]
])

predictions = model.predict(students)

print("\nStudent Predictions")
print("-"*50)

for i in range(len(students)):

    result = "PASS" if predictions[i] >= 0.5 else "FAIL"

    print(f"Student {i+1}")
    print(f"Study Hours      : {students[i][0]}")
    print(f"Attendance       : {students[i][1]}%")
    print(f"Assignment Score : {students[i][2]}")
    print(f"Prediction Score : {predictions[i][0]:.3f}")
    print(f"Result           : {result}")
    print("-"*50)
```
