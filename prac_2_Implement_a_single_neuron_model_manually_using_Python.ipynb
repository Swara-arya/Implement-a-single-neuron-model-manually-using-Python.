Step 1: Import NumPy

```python
import numpy as np
```

Step 2: Define Input Features

We will use three features:

Number of suspicious words
Number of links in the email
Number of capitalized words

```python
# Input Features
# [Suspicious Words, Links, Capital Words]
email = np.array([6, 3, 8])

print("Email Features:")
print(email)
```

Step 3: Initialize Weights and Bias

```python
# Weights assigned to each feature
weights = np.array([0.45, 0.35, 0.25])

# Bias
bias = -2

print("\nWeights:", weights)
print("Bias:", bias)
```

Step 4: Calculate Weighted Sum

```python
weighted_sum = np.dot(email, weights) + bias

print("\nWeighted Sum =", weighted_sum)
```

Step 5: Apply Sigmoid Activation Function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

output = sigmoid(weighted_sum)

print("\nNeuron Output =", output)
```

Step 6: Predict Spam or Not Spam

```python
if output >= 0.5:
    print("\nPrediction: Spam Email")
else:
    print("\nPrediction: Not Spam")
```

Step 7: Test Multiple Emails

```python
test_emails = np.array([
    [6, 3, 8],   # Spam
    [1, 0, 1],   # Not Spam
    [5, 2, 6],   # Spam
    [0, 0, 2],   # Not Spam
    [4, 1, 5]    # Spam
])

print("\nTesting Different Emails")
print("-" * 55)

for email in test_emails:

    weighted_sum = np.dot(email, weights) + bias
    output = sigmoid(weighted_sum)

    prediction = "Spam" if output >= 0.5 else "Not Spam"

    print(f"Email Features : {email}")
    print(f"Weighted Sum   : {weighted_sum:.2f}")
    print(f"Neuron Output  : {output:.3f}")
    print(f"Prediction     : {prediction}")
    print("-" * 55)
```
