# **Application: Fruit Classification**

The model predicts the type of fruit based on:

Weight (grams)
Sweetness (1–10)
Color Intensity (1–10)
Classes

Apple → 0

Banana → 1

Orange → 2

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical
```

```python
# Features:
# Weight(g), Sweetness, Color Intensity

X = np.array([
    [170,6,8],
    [180,7,9],
    [160,6,7],
    [120,9,4],
    [110,10,3],
    [130,8,5],
    [150,7,6],
    [155,8,7],
    [145,7,8]
],dtype=float)

# Classes
# Apple=0
# Banana=1
# Orange=2

y=np.array([
0,0,0,
1,1,1,
2,2,2
])

# One-Hot Encoding
y=to_categorical(y,num_classes=3)

print("Dataset Loaded Successfully")
```

Build the MLP Model

```python
model=Sequential()

model.add(Dense(8,input_dim=3,activation='relu'))

model.add(Dense(6,activation='relu'))

model.add(Dense(3,activation='softmax'))
```

Compile the Model

```python
model.compile(
optimizer='adam',
loss='categorical_crossentropy',
metrics=['accuracy']
)
```

Train the Model

```python
history=model.fit(
X,
y,
epochs=100,
verbose=1
)
```

 Evaluate Model

```python
loss,accuracy=model.evaluate(X,y)

print("\nAccuracy:",accuracy)
```

Predict New Fruits

```python
test=np.array([
[175,7,8],
[115,9,4],
[148,8,7]
])

prediction=model.predict(test)

classes=["Apple","Banana","Orange"]

print("\nPredictions")
print("-"*40)

for i in range(len(test)):

    predicted=np.argmax(prediction[i])

    print("Input:",test[i])
    print("Predicted Fruit:",classes[predicted])
    print("Prediction Probabilities:",prediction[i])
    print("-"*40)
```
